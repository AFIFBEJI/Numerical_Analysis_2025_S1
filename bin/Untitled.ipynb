def matrice(X,p):
    n = len(X)
    A = np.zeros((n, p+1))
    for i in range (n):
        for j in range (p+1):
            A[i,j]=X[i]**j
    return A

```python
import numpy as np

def regression(X, Y, p):
    A = np.vander(X, p+1, increasing=True)
    # REMPLACEZ la ligne suivante :
    Lambd = np.linalg.pinv(A) @ Y  # ← UTILISEZ CELLE-CI
    return Lambd
```

```python
# Testons avec vos données
X = np.array([0, 1, 2, 3, 4, 5])
Y = np.array([1, 1.5, 2.5, 3.5, 5.0, 7.0])
p = 3

Lambda = regression(X, Y, p)
print(Lambda)
```
